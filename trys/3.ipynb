```python
from collections import deque

# تعریف گراف مسیرها با توجه به نقشه داده شده
graph = {
    'A': {'B': 'red'},
    'B': {'C': 'red', 'E': 'blue'},
    'C': {'D': 'red', 'F': 'blue'},
    'D': {'I': 'green'},
    'E': {'F': 'blue', 'K': 'green'},
    'F': {'G': 'blue', 'L': 'green'},
    'G': {'H': 'blue', 'M': 'green'},
    'H': {'I': 'blue', 'N': 'green'},
    'I': {'J': 'red'},
    'K': {'L': 'green', 'P': 'red'},
    'L': {'M': 'green', 'Q': 'red'},
    'M': {'N': 'green', 'R': 'red'},
    'N': {'T': 'red'},
    'P': {'Q': 'red', 'U': 'blue'},
    'Q': {'R': 'red', 'V': 'blue'},
    'R': {'S': 'red', 'W': 'blue'},
    'S': {'T': 'red', 'X': 'blue'},
    'T': {'Y': 'blue'},
    'U': {'V': 'blue', 'a': 'green'},
    'V': {'W': 'blue', 'b': 'green'},
    'W': {'X': 'blue', 'c': 'green'},
    'X': {'Y': 'blue', 'd': 'green'},
    'Y': {'e': 'green'},
    'a': {'b': 'green', 'f': 'red'},
    'b': {'c': 'green', 'g': 'red'},
    'c': {'d': 'green', 'h': 'red'},
    'd': {'e': 'green', 'i': 'red'},
    'f': {'g': 'red'},
    'g': {'h': 'red'},
    'h': {'i': 'red'},
    'i': {}
}

# تعریف نوع هر خط (برای بررسی انتقال رایگان)
line_types = {
    'red': 'cable',
    'blue': 'horse',
    'green': 'bus'
}

# تابع برای پیدا کردن مسیر با استفاده از BFS
def find_path_with_transfers(graph, start, goal):
    queue = deque([(start, [], None)])  # (نقطه فعلی، مسیر طی‌شده، خط فعلی)
    visited = set()

    while queue:
        current, path, current_line = queue.popleft()

        # اگر به هدف رسیدیم
        if current == goal:
            return path + [current]

        # جلوگیری از بررسی نقاط تکراری
        if (current, current_line) in visited:
            continue
        visited.add((current, current_line))

        # بررسی مسیرهای ممکن از نقطه فعلی
        for neighbor, line in graph[current].items():
            # انتقال رایگان یا ادامه در همان خط
            if current_line is None or line_types[line] == line_types[current_line]:
                queue.append((neighbor, path + [current], line))

    return None

# پیدا کردن مسیر از Startsburg به Endenville
start = 'A'
goal = 'i'
path = find_path_with_transfers(graph, start, goal)

if path:
    print("Path from Startsburg to Endenville:")
    print(" -> ".join(path))
else:
    print("No path found.")
```
