```python
def find_all_paths(graph, start_node):
    all_paths = []

    def dfs(path, color=None, transport=None):
        current_node = path[-1]
        for neighbor in graph.neighbors(current_node):
            edge_data = graph.get_edge_data(current_node, neighbor)
            if neighbor not in path:
                if color is None or edge_data['color'] == color or edge_data['transport'] == transport:
                    dfs(path + [neighbor], edge_data['color'], edge_data['transport'])
        all_paths.append(path)

    dfs([start_node])
    return all_paths
```
